```python
import numpy as np
from power_calc.continuous import ContinuousEndpoint
from power_calc.binary import BinaryEndpoint
from power_calc.survival import SurvivalEndpoint
```

```python
trial_continuous = ContinuousEndpoint(outcome_sd=1.0)  # sd=1.0 to match Cohen's d directly

n_needed = trial_continuous.closed_form_sample_size(treatment_effect=0.5, power=0.8)
print(n_needed)  
```

```python
trial_binary = BinaryEndpoint()

n_needed = trial_binary.closed_form_sample_size(p1=0.30, p2=0.176, power=0.8)
print(n_needed)
print(trial_binary.closed_form_power(n_needed, p1=0.30, p2=0.176))  # should land at 0.8
```

```python
trial_survival = SurvivalEndpoint(accrual_period=6.0, follow_up_period=18.0)

# HR=2 and HR=0.5 require the identical number of events
print(trial_survival.closed_form_events_needed(hazard_ratio=2.0))
print(trial_survival.closed_form_events_needed(hazard_ratio=0.5))

result = trial_survival.closed_form_sample_size(hazard_ratio=0.6, control_median_survival=12.0, power=0.8)
print(result)  
```
